# 1. Import Necessary Libraries

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
```

# 2. Import Dataset

```python
claimants_data = pd.read_csv(r"C:\Users\gthangar\OneDrive - Nokia\1830_Automation\ML\Task\Machine_Learning\03-Logistic_Regression\claimants.csv")
claimants_data
```

# 4. Data Preparation

### 4.1 Data Cleaning

```python
claimants_data.dropna(inplace=True)
```

```python
claimants_data.shape
```

### 4.2 Separate your Input and Output

```python
X = claimants_data.drop(labels = ['CASENUM','ATTORNEY'], axis=1)
y = claimants_data['ATTORNEY']
```

# 5. Model Building

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size= 0.80, random_state=42)
print("Training Data",len(X_train),len(y_train))
print("Testing Data",len(X_test),len(y_test))
```

```python
logistic_model = LogisticRegression()
logistic_model
```

# 6. Model Training

```python
logistic_model.fit(X_train,y_train)
```

# 7. Model Testing

### 7.1 Model Testing for Trainig Data 

```python
y_pred_train = logistic_model.predict(X_train)
y_pred_train
```

### 7.2 Model Testing for Test Data

```python
y_pred_test = logistic_model.predict(X_test)
y_pred_test
```

# 8. Model Evaluation

```python
from sklearn.metrics import accuracy_score,classification_report,precision_score,recall_score,roc_auc_score,roc_curve,confusion_matrix
```

### 8.1 Training Data's Evaluation Metrices

```python
print(classification_report(y_true=y_train,y_pred=y_pred_train))
```

```python
confusion_matrix(y_true=y_train,y_pred=y_pred_train)
```

```python
accuracy_score(y_true=y_train,y_pred=y_pred_train)
```

### ROC Curve - Receiver Operating Characteristics

##### AUC - Area Under this Curve

```python
# Get predicted probabilities for Class 1
y_prob_train = logistic_model.predict_proba(X_train)[:,1]

# Calculate AUC
auc = roc_auc_score(y_train, y_prob_train)
print("AUC:", auc)

# Calculate FPR, TPR and thresholds
fpr, tpr, thresholds = roc_curve(y_true=y_train, y_score=y_prob_train)

# Plot ROC curve
plt.plot(fpr,tpr,color='red')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.show()
```

### 8.2 Test Data's Evaluation Metrices

```python
print(classification_report(y_true=y_test,y_pred=y_pred_test))
```

```python
confusion_matrix(y_true=y_test,y_pred=y_pred_test)
```

```python
accuracy_score(y_true=y_test,y_pred=y_pred_test)
```

```python
y_prob_test = logistic_model.predict_proba(X_test)[:,1]

# Calculate AUC
auc = roc_auc_score(y_true=y_test,y_score=y_prob_test)
print("AUC:", auc)

# Calculate FPR, TPR and thresholds
fpr, tpr, thresholds = roc_curve(y_true=y_test, y_score=y_prob_test)

# Plot ROC curve
plt.plot(fpr,tpr,color='red')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.show()
```
